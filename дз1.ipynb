## Задача 1.8

В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

```python
from math import comb

film = 10
nomination = 5

a = film**nomination
b = comb(film + nomination -1, nomination)

print(f'а) {a}')
print(f'б) {b}')
```

## Задача 1.9

Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

```python
from math import factorial,prod

n = 7
povtor = [3, 2, 2]

res = factorial(n) // prod(factorial(i) for i in povtor)

res
```

## Задача 1.10

Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребёнок берёт карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) ТОР; б) ТЕОРИЯ?

Событие A - получится слово ТОР

Собитие B - получится слово ТЕОРИЯ

```python
from math import perm, factorial

n_all = 6
n_tor = 3
a = perm(n_all, n_tor)
p_a = 1/a

b = factorial(6)
p_b = 1/b

print(f'а) {p_a}')
print(f'б) {p_b}')
```

## Задача 1.16

За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

Событие A - никакие два лица одного пола не сядут рядом

Событие B - супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

```python
from math import factorial

m1 = 2 * (factorial(5)**2)
p_a = m1 / factorial(10)

m2 = factorial(5) * (2**5)
p_b = m2 / factorial(10)

print(f'а) {p_a}')
print(f'б) {p_b}')
```

## Задача 1.17

В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятность того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

Событие A - пассажиры попали в два купе

Событие B - пассажиры попали в семь купе

Событие C - пассажиры попали в три купе

```python
from math import comb

n = comb(36, 7)

a = comb(9, 2) * comb(8, 7)
p_a = a/n

b = comb(9, 7) * (4**7)
p_b = b/n

_421 = 6 * (comb(4, 4) * comb(4, 2) * comb(4,1))
_331 = 3 * (comb(4, 3) ** 2 * comb(4, 1))
_322 = 3 * (comb(4, 3) * comb(4, 2) ** 2)
c = comb(9, 3) * (_421 + _331 + _322)
p_c = c/n

print(f'а) {p_a:.7f}')
print(f'б) {p_b:.7f}')
print(f'в) {p_c:.7f}')
```

в) нечётные;

## Задача 1.48

В старинной игре в кости необходимо было для выигрыша получить при бросании трёх игральных костей сумму очков, превосходящую 10. Найти вероятности:

а) выпадения 11 очков;

б) выигрыша.

Событие A - при бросании трёх игральных костей выпадет 11 очков

Событие B - при бросании трёх игральных костей сумма очков превосходит 10

```python
from itertools import product

ishod = list(product(range(1,7), repeat=3))
sposob = len(ishod)

ishod_11 = 0
for i in ishod:
    if sum(i)==11:
        ishod_11 += 1

p_a = ishod_11/sposob

ishod_10 = 0
for i in ishod:
    if sum(i)>10:
        ishod_10 += 1

p_b = ishod_10/sposob

print(f'а) {p_a}')
print(f'б) {p_b}')
```
